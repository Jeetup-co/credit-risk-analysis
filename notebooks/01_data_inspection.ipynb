# Phase 2 — Data Understanding Summary

## Dataset Overview

- Dataset: `LC_loans_granting_model_dataset.csv`
- Rows: 1,347,681
- Columns: 15
- Date range: Apr-2008 to Sep-2018
- Target variable: `Default`

## Target Variable

`Default` is a binary variable:

- `0` → Loan did not default
- `1` → Loan defaulted

Target distribution:

- Default = 0: 1,078,432 (80.02%)
- Default = 1: 269,249 (19.98%)

The target is moderately imbalanced, so class distribution should be considered during later modeling and evaluation.

## Important Features

### Financial Features
- `revenue` — borrower revenue/income-related numerical feature
- `dti_n` — debt-to-income related feature
- `loan_amnt` — loan amount
- `fico_n` — FICO credit score

### Employment Features
- `experience_c` — binary experience indicator
- `emp_length` — employment length category

### Loan Features
- `purpose` — purpose of the loan
- `issue_d` — loan issue month/year

### Borrower / Location Features
- `home_ownership_n` — home ownership category
- `addr_state` — borrower's state
- `zip_code` — ZIP code

### Text Features
- `title` — loan title
- `desc` — loan description

### Identifier
- `id` — unique loan identifier

## Data Quality Observations

- No duplicate rows were found.
- No duplicate loan IDs were found.
- `zip_code` contains 1 missing value.
- `title` contains 16,657 missing values.
- `desc` contains 1,228,582 missing values (~91.16%).
- `revenue` has a large range and potential outliers.
- `dti_n` contains suspicious high values, including `999`.
- The `dti_n = 999` values will **not** be modified during Data Understanding. They will be investigated and handled during **Phase 4 — Data Quality**.

## Initial Business Questions

The project will investigate:

1. Which borrower characteristics are associated with loan default?
2. Does lower FICO score correspond to higher default risk?
3. Does a higher debt-to-income ratio correspond to higher default risk?
4. Does loan amount influence default probability?
5. Does borrower employment length affect default risk?
6. Does home ownership status affect default risk?
7. Which loan purposes have higher default rates?
8. Can borrower and loan characteristics be used to predict loan default?

## Phase 2 Conclusion

The dataset is sufficiently large for statistical analysis and machine learning.

The main target is clearly defined, important numerical and categorical variables have been identified, and initial data-quality concerns have been documented.

Further cleaning and transformation will be performed in later phases rather than during Data Understanding.
