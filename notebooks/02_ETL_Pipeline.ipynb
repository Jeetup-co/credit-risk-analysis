# Phase 3 — ETL Pipeline

## Extract

The Extract stage loads the raw Lending Club dataset from the `data/raw/` directory into a pandas DataFrame.

The raw dataset will not be modified during extraction.

```python
import pandas as pd

raw_path = '../data/raw/LC_loans_granting_model_dataset.csv'

df_raw = pd.read_csv(raw_path)

df_raw.shape
```

```python
df_raw.columns.tolist()
```

## Transform

The Transform stage applies controlled changes required to prepare the extracted data for downstream analysis.

Transformations will be performed without modifying the original raw dataset.


```python
df_raw['issue_d'].head()
```

```python
df_raw['issue_d'].dtype
```

```python
df_etl = df_raw.copy()

df_etl['issue_d'] = pd.to_datetime(
    df_etl['issue_d'],
    format = "%b-%Y"
)

df_etl['issue_d'].dtype
```

```python
df_etl['issue_d'].head()
```

```python
df_etl['issue_d'].min(),df_etl['issue_d'].max()
```

```python
df_etl = df_raw.copy()

df_etl["issue_d"] = pd.to_datetime(
    df_raw["issue_d"],
    format="%b-%Y"
)

df_etl["issue_d"].min(), df_etl["issue_d"].max()
```

```python
print("RAW:")
print(df_raw["issue_d"].head())

print("\nTRANSFORMED:")
print(df_etl["issue_d"].head())
```

## Load

The Load stage saves the transformed dataset into the processed data directory.

The original raw dataset remains unchanged in `data/raw/`, while the transformed dataset is stored separately in `data/processed/`.

This separation ensures reproducibility and allows downstream analysis to use a controlled, transformed version of the data.

```python
from pathlib import Path

processed_path = Path('../data/processed')

processed_path.mkdir(parents=True, exist_ok=True)
```

```python
processed_path = Path('../data/processed/LC_loans_granting_model_dataset_processed.csv')

df_etl.to_csv(processed_path, index=False)

print("Processed data saved to: ")
print(processed_path)

print("\nProcessed data shape: ")
print(df_etl.shape)
```
